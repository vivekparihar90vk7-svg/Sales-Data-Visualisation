# Sales Data Visualization and Analysis

### Objective

The main objective of this project is to explore sales data,
identify useful patterns and trends, and communicate the findings
through clear and meaningful visualizations.

### Tools Used

- Python
- Pandas
- Matplotlib
- Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Project setup completed successfully.")
```

```python
file_path = "../data/raw/Sample - Superstore.csv"

sales_data = pd.read_csv(file_path, encoding="latin1")

print("Dataset loaded successfully!")
```

```python
print("Rows and columns:", sales_data.shape)
sales_data.head()
```

```python
sales_data.head()
```

```python
print("Dataset Information:")
sales_data.info()
```

```python
print(sales_data.isna().sum())
```

```python
print("Missing values in each column:")
print(sales_data.isnull().sum())
```

```python
sales_data["Order Date"] = pd.to_datetime(sales_data["Order Date"])
sales_data["Ship Date"] = pd.to_datetime(sales_data["Ship Date"])

print("Date columns converted successfully!")
```

```python
print(sales_data[["Order Date", "Ship Date"]].dtypes)
```

```python
duplicate_count = sales_data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

```

```python
print(sales_data[["Sales", "Quantity", "Discount", "Profit"]].describe())
```

```python
cleaned_file_path = "../data/cleaned/superstore_cleaned.csv"

sales_data.to_csv(cleaned_file_path, index=False)

print("Cleaned dataset saved successfully!")
```

```python
print("Total Sales:", sales_data["Sales"].sum())
print("Total Profit:", sales_data["Profit"].sum())
print("Total Quantity Sold:", sales_data["Quantity"].sum())
print("Average Discount:", sales_data["Discount"].mean())
```

```python
category_sales = sales_data.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig("../visualisation/sales_by_category.png", dpi=300)
plt.show()
```

```python
category_profit = sales_data.groupby("Category")["Profit"].sum().sort_values(ascending=False)

print(category_profit)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(category_profit.index, category_profit.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../visualisation/profit_by_category.png", dpi=300)
plt.show()
```

```python
region_analysis = sales_data.groupby("Region")[["Sales", "Profit"]].sum()

print(region_analysis)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(region_analysis.index, region_analysis["Sales"])

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig("../visualisation/sales_by_region.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(region_analysis.index, region_analysis["Profit"])

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../visualisation/profit_by_region.png", dpi=300)
plt.show()
```

```python
sales_data["Month"] = sales_data["Order Date"].dt.to_period("M")

monthly_sales = sales_data.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python

plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../visualisation/monthly_sales_trend.png", dpi=300)
plt.show()
```

```python
subcategory_sales = sales_data.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

print(subcategory_sales)
```

```python

plt.figure(figsize=(10, 7))

plt.barh(subcategory_sales.index, subcategory_sales.values)

plt.title("Sales by Sub-Category")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("../visualisation/sales_by_subcategory.png", dpi=300)
plt.show()
```

```python
subcategory_profit = sales_data.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

print(subcategory_profit)
```

```python
plt.figure(figsize=(10, 7))

plt.barh(subcategory_profit.index, subcategory_profit.values)

plt.title("Profit by Sub-Category")
plt.xlabel("Profit")
plt.ylabel("Sub-Category")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.savefig("../visualisation/profit_by_subcategory.png", dpi=300)
plt.show()
```

```python
segment_sales = sales_data.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

print(segment_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(segment_sales.index, segment_sales.values)

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig("../visualisation/sales_by_segment.png", dpi=300)
plt.show()
```

```python
segment_profit = sales_data.groupby("Segment")["Profit"].sum().sort_values(ascending=False)

print(segment_profit)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(segment_profit.index, segment_profit.values)

plt.title("Profit by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../visualisation/profit_by_segment.png", dpi=300)
plt.show()
```

```python
discount_profit = sales_data.groupby("Discount")["Profit"].mean()

print(discount_profit)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(discount_profit.index, discount_profit.values, marker="o")

plt.title("Discount vs Average Profit")
plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.tight_layout()

plt.savefig("../visualisation/discount_vs_profit.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(sales_data["Sales"], sales_data["Profit"], alpha=0.5)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.tight_layout()

plt.savefig("../visualisation/sales_vs_profit.png", dpi=300)
plt.show()
```

## Key Insights

### Sales Performance
- Technology generated the highest total sales among the three categories.
- The West region recorded the highest total sales.
- Consumer customers contributed the highest sales among all customer segments.
- Phones and Chairs were the top-selling sub-categories.

### Profit Performance
- Technology generated the highest total profit among the categories.
- The West region generated the highest total profit.
- Copiers generated the highest profit among the sub-categories.
- Tables recorded the highest loss among the sub-categories.

### Discount Analysis
- Higher discount levels were generally associated with lower average profit.
- Discounts of 30% or more showed negative average profit in this dataset.

### Overall Observation
The analysis shows that high sales do not always guarantee high profitability. Product category, sub-category, region, customer segment, and discount level all have an impact on business performance.

## Conclusion

This project analyzed the Superstore sales dataset using Python and Pandas.

The analysis explored sales, profit, customer segments, regions, product categories, sub-categories, and the relationship between discounts and profit.

The visualizations helped identify important business patterns, such as the strong performance of the Technology category and West region, while also showing that higher discounts can negatively affect profitability.

Overall, this project demonstrates how data visualization can be used to understand business performance and extract meaningful insights from raw data.
